```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('netflix_titles.csv')
display(df.head(11))
```

```python
print("Shape:", df.shape)
```

```python
print(df.isnull().sum())
```

```python
df["director"].fillna("Unknown",inplace=True)
df["cast"].fillna("Not Available",inplace=True)
df["country"].fillna("Unknown",inplace=True)
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df["country"].value_counts().head(5).plot(kind="bar")
plt.title("Top 5 Countries")
plt.xlabel("country")
plt.ylabel("count")
plt.show()
```

```python
df["type"].value_counts().plot(kind="bar")
plt.title("Movie vs TV Shows")
plt.xlabel("type")
plt.ylabel("count")
plt.show()
```

**CONCLUSION**

This project shows that Netflix focuses more on movies than TV shows and has rapidly expanded its content library in recent years. Most content comes from a few major countries, especially the United States, and the majority of titles are aimed at mature audiences. Data visualization helps easily understand trends and patterns in streaming platforms.
